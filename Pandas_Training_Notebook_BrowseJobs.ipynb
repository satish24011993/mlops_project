# Chapter 1 · Why Pandas — The Four Data Problems

**Pandas is a framework used to clean data.** Raw data arriving from a business is almost never usable as-is. There are four problems we fix, every single time:

| # | Problem | Example |
|---|---|---|
| 1 | **Missing data** | salary is `null`, city is `unknown` |
| 2 | **Duplicate data** | the same customer row appears three times |
| 3 | **Wrong data type** | salary stored as `'180000'` (a string) instead of a number |
| 4 | **Wrong data** | `city = 'male'`, a phone number sitting in the salary column |

### Look at this raw table — every row has a problem
```
customername | age | salary       | city
a            | 16  | null         | unknown      ← missing data
b            | 18  | 150000       | bangalore    ← clean
c            | 21  | 180000       | chennai      ← clean
e            | 25  | 180000       | bangalore
e            | 25  | '180000'     | bangalore    ← duplicate + wrong TYPE (string)
e            | 25  | 6363402444   | male         ← wrong data in BOTH columns
```
If you group this by city to report total sales, your numbers are **wrong** — that is why cleaning comes first.

**Why we clean:** data analysis / reporting · prediction modelling. Garbage in → garbage dashboards.

### Array vs List (the vocabulary question)
- **Array** — *homogeneous*: every element the same type.
- **List** — *heterogeneous*: can hold any mix of types.

# Chapter 6 · Cleaning 1 — Duplicate Data

Duplicates inflate every total in your dashboard. Find them, then drop them.

⚠️ **The `inplace` trap:** `df.drop_duplicates()` returns a *new* DataFrame and leaves `df` untouched. Either assign the result or pass `inplace=True`.


````python
df = pd.read_csv('data.csv')
print('rows before:', len(df))

# Find them first
print('duplicate rows:', df.duplicated().sum())
df[df.duplicated(keep=False)]      # show every copy
````

````python
# ❌ This does NOT change df — a classic mistake
df.drop_duplicates()
print('still:', len(df))

# ✅ Either of these works
newdf = df.drop_duplicates()       # assign the result
df.drop_duplicates(inplace=True)   # or modify in place
print('after:', len(df))
````

````python
# Drop duplicates based on SPECIFIC columns only
df2 = pd.read_csv('data.csv')
df2.drop_duplicates(subset=['Duration', 'Pulse'], keep='first').head()
````

# Chapter 7 · Cleaning 2 — Missing Data

Two strategies:

| Strategy | Command | When to use |
|---|---|---|
| **Delete** the rows | `dropna()` | Few nulls, plenty of data |
| **Fill** the gaps | `fillna(value)` | Deleting would lose too much |

### What do we fill WITH?
1. A **fixed value** — `fillna(130)`
2. A value **provided by the business** — ask, don't guess
3. A **statistic**: **mean** (average) · **median** (middle value) · **mode** (most frequent)

**The BrowseJobs rule of thumb:** numeric → mean or median · date → a sentinel like `01/01/1900` · string → `'unknown'`.


````python
df = pd.read_csv('data.csv')

# Count nulls per column — do this BEFORE deciding anything
df.isnull().sum()
````

````python
# --- Strategy 1: DELETE the null rows ---
newdf = df.dropna()
print('original:', len(df), ' after dropna:', len(newdf))
newdf.info()
````

````python
# --- Strategy 2a: fill EVERYTHING with a fixed value ---
df = pd.read_csv('data.csv')
newdf = df.fillna(130)
newdf.isnull().sum()
````

````python
# --- Strategy 2b: fill ONE column only (much more common) ---
df = pd.read_csv('data.csv')
df['Calories'] = df['Calories'].fillna(130)
df.isnull().sum()
````

````python
# --- Strategy 2c: mean / median / mode ---
df = pd.read_csv('data.csv')

x = df['Calories'].mean()      # average
y = df['Pulse'].median()       # middle value
z = df['Duration'].mode()[0]   # most frequent — mode() returns a Series, take [0]!

print('mean Calories:', round(x, 2), '| median Pulse:', y, '| mode Duration:', z)

df['Calories'] = df['Calories'].fillna(x)
df.isnull().sum()
````

### ⚠️ Two traps students always hit
1. **`mode()` returns a Series**, not a single value — you must index it: `df['col'].mode()[0]`. Passing the Series straight into `fillna()` silently does nothing useful.
2. **`inplace=True` on a column slice is deprecated** in modern pandas. Prefer the assignment form:
   `df['col'] = df['col'].fillna(x)` instead of `df['col'].fillna(x, inplace=True)`.

### 💬 Interview Corner — mean vs median
Use the **median** when the column has outliers (one salary of ₹5 crore drags the mean up but barely moves the median). Use the **mode** for categorical columns.

# Chapter 8 · Cleaning 3 — Wrong Data Types & Wrong Data

### Wrong data type
A date stored as a string can't be sorted, filtered by month, or used in a time series. Fix it with `pd.to_datetime()`.

### Wrong data
Data that is *valid* as a type but *impossible* in reality — a 450-minute workout, a negative salary, `city = 'male'`. Two options: **correct it** to a boundary value, or **delete** the row.


````python
import pandas as pd
df = pd.read_csv('data.csv')
df.info()      # look at Dtype: Date is 'object' = a string!
````

````python
# Convert to a real datetime.
# errors='coerce' turns anything unparseable into NaT instead of crashing.
df['Date'] = pd.to_datetime(df['Date'], errors='coerce', format='mixed')
df.info()      # Date is now datetime64
````

````python
# Now fill the resulting missing dates with the business sentinel
df['Date'] = df['Date'].fillna(pd.Timestamp('1900-01-01'))
df.head(3)
````

````python
# Reformat the display: dd-mm-yyyy
df['Date_display'] = df['Date'].dt.strftime('%d-%m-%Y')
df[['Date', 'Date_display']].head()
````

````python
# --- WRONG DATA: fix out-of-range values ---
# Row 23 has a Duration of 450 minutes — nobody works out for 7.5 hours.
print('before:', df['Duration'].max())

for i in df.index:
    if df.loc[i, 'Duration'] > 120:
        df.loc[i, 'Duration'] = 120        # cap it at a sensible boundary

print('after :', df['Duration'].max())
````

````python
# The vectorised version — faster and more 'pandas-native' than looping
df2 = pd.read_csv('data.csv')
df2.loc[df2['Duration'] > 120, 'Duration'] = 120
print(df2['Duration'].max())
````

````python
# Or DELETE the bad rows entirely
df3 = pd.read_csv('data.csv')
df3 = df3[df3['Duration'] <= 120]
print('rows kept:', len(df3))
````

# Chapter 9 · Data Validation Framework

Validation runs **before** transformation (is the file even usable?) and **after** transformation (did we break anything?). This checklist is what separates a data engineer from someone who "knows pandas".

| # | Validation | What you check |
|---|---|---|
| 1 | **Naming convention** | `finance_01012026.csv` — not `f.csv` |
| 2 | **File type** | Is it actually a CSV, not an Excel file renamed? |
| 3 | **File size** | A 0 KB or suspiciously small file = upstream failure |
| 4 | **Row count** | Actual vs expected number of rows |
| 5 | **Schema** | Expected columns vs actual columns |
| 6 | **Data type** | Expected dtypes vs actual dtypes |
| 7 | **Missing data** | Count of nulls per column |
| 8 | **Wrong data** | Values outside the expected range/domain |
| 9 | **Duplicates** | Count of duplicate rows |


````python
import pandas as pd, os

def validate(path, expected_cols, expected_min_rows=1):
    """A minimal validation harness you can drop into any pipeline."""
    report = {}
    report['file_exists']  = os.path.exists(path)
    report['file_size_kb'] = round(os.path.getsize(path) / 1024, 2)

    df = pd.read_csv(path)
    report['row_count']       = len(df)
    report['row_count_ok']    = len(df) >= expected_min_rows
    report['schema_ok']       = list(df.columns) == expected_cols
    report['missing_columns'] = set(expected_cols) - set(df.columns)
    report['null_counts']     = df.isnull().sum().to_dict()
    report['duplicate_rows']  = int(df.duplicated().sum())
    report['dtypes']          = df.dtypes.astype(str).to_dict()
    return report

for k, v in validate('data.csv',
                     ['Duration','Date','Pulse','Maxpulse','Calories'],
                     expected_min_rows=25).items():
    print(f'{k:18} : {v}')
````

# Chapter 10 · Filtering & Creating New Columns

### The filtering syntax trap ⚠️
When combining conditions you must use:
- `&` for AND, `|` for OR (**not** `and` / `or`)
- **Round brackets around every condition** — `&` binds tighter than `>`

```python
df[(df['age'] > 30) & (df['city'] == 'bangalore')]     # ✅
df[df['age'] > 30 & df['city'] == 'bangalore']         # ❌ error
```


````python
import pandas as pd
df = pd.read_csv('data.csv')

# Single condition
df[df['Calories'] > 350].head()
````

````python
# Two conditions with & — note the brackets around EACH one
df[(df['Calories'] > 300) & (df['Pulse'] < 105)].head()
````

````python
# OR with |
df[(df['Duration'] == 45) | (df['Pulse'] > 120)].head()
````

````python
# isin() — the pandas equivalent of SQL's IN
df[df['Duration'].isin([30, 45])].head()
````

````python
# Creating new columns — plain arithmetic across columns
df['intensity']    = df['Calories'] / df['Duration']
df['pulse_reserve'] = df['Maxpulse'] - df['Pulse']
df[['Duration','Calories','intensity','pulse_reserve']].head()
````

````python
# Conditional column — the pandas version of a SQL CASE statement
import numpy as np
df['effort'] = np.where(df['Calories'] > 350, 'high',
               np.where(df['Calories'] > 250, 'medium', 'low'))
df[['Calories','effort']].head(8)
````

# Chapter 11 · Merging DataFrames — the Joins

`pd.merge()` is the pandas version of a SQL JOIN. The `how=` parameter picks the type.

| `how=` | Result |
|---|---|
| `'inner'` | Only **matching** rows from both |
| `'left'` | All from the **left** + matches from the right (nulls where no match) |
| `'right'` | All from the **right** + matches from the left |
| `'outer'` | **Everything** from both tables |
| `'cross'` | Cartesian product — every row × every row |


````python
import pandas as pd

df1 = pd.DataFrame({'id': [1, 2, 3, 4], 'Name': ['krish', 'ajay', 'adveer', 'anuj']})
df2 = pd.DataFrame({'id': [3, 4, 5], 'age': [25, 28, 22]})
print(df1); print(); print(df2)
````

````python
# INNER — only ids 3 and 4 exist in both
pd.merge(df1, df2, on='id', how='inner')
````

````python
# LEFT — all four names; ages are NaN where there's no match
pd.merge(df1, df2, on='id', how='left')
````

````python
# RIGHT — all three ages; id 5 has no name
pd.merge(df1, df2, on='id', how='right')
````

````python
# OUTER (full outer join) — everything from both
pd.merge(df1, df2, on='id', how='outer')
````

````python
# When the key columns have DIFFERENT names, use left_on / right_on
customers = pd.DataFrame({'id': [1, 2, 3], 'name': ['a', 'b', 'c']})
orders    = pd.DataFrame({'customerid': [1, 1, 3], 'amount': [100, 250, 90]})

pd.merge(customers, orders, left_on='id', right_on='customerid', how='left')
````

````python
# Merging on MULTIPLE keys — pass a list
a = pd.DataFrame({'city':['bangalore','ajmer','kolkata','delhi'],
                  'id':[1,2,3,4], 'Name':['krish','ajay','adveer','anuj']})
b = pd.DataFrame({'city':['kolkata','ajmer','bangalore','delhi'],
                  'id':[3,4,1,2], 'age':[25,28,22,24]})

print(pd.merge(a, b, on=['id'], how='inner'))          # city_x / city_y suffixes appear
print()
print(pd.merge(a, b, on=['id','city'], how='inner'))   # matches BOTH keys — fewer rows
````

````python
# Chaining THREE tables: merge two, then merge the result with the third
t1 = pd.DataFrame({'id':[1,2,3], 'Name':['john','Anna','Peter']})
t2 = pd.DataFrame({'id':[3,4,5], 'age':[25,28,22]})
t3 = pd.DataFrame({'id':[1,2,3], 'country':['USA','UK','India']})

output = pd.merge(t1, t2, on='id', how='inner')
newdf  = pd.merge(output, t3, on='id', how='inner')
newdf
````

````python
# concat — STACK dataframes instead of joining them (rows on top of rows)
pd.concat([t1, pd.DataFrame({'id':[9], 'Name':['zara']})], ignore_index=True)
````

### 💬 Interview Corner — merge vs concat vs join
- **`merge`** — combine on **key columns** (SQL JOIN).
- **`concat`** — **stack** dataframes vertically (or side by side with `axis=1`); no key needed.
- **`join`** — merges on the **index** by default; a convenience wrapper around merge.

# Chapter 12 · Reshaping — pivot & json_normalize

### pivot — turn rows into columns
Long format (one row per date-city) → wide format (dates as rows, cities as columns).

### json_normalize — flatten nested JSON
APIs return nested JSON. `json_normalize` flattens `address.city` into a proper column.


````python
import pandas as pd

df = pd.DataFrame({
    'date': ['2021-01-01', '2021-01-01', '2021-01-02', '2021-01-02'],
    'city': ['newyork', 'los angeles', 'newyork', 'los angeles'],
    'temperature': [32, 75, 34, 77]})
df
````

````python
# index = the rows | columns = what becomes column headers | values = the cells
newdf = df.pivot(index='date', columns='city', values='temperature')
newdf
````

````python
# Pivot MULTIPLE value columns at once → a multi-level column header
data = {
    'Date': ['2023-01-01','2023-01-01','2023-01-02','2023-01-02'],
    'City': ['New York','Los Angeles','New York','Los Angeles'],
    'Temperature': [32, 75, 34, 77],
    'Humidity': [80, 60, 70, 65],
    'Wind Speed': [10, 15, 12, 14]
}
df = pd.DataFrame(data)
df
````

````python
newdf = df.pivot(index='Date', columns='City', values=['Temperature','Humidity','Wind Speed'])
print(newdf)
````

````python
# pivot_table — like pivot but AGGREGATES duplicates instead of erroring
df.pivot_table(index='City', values=['Temperature','Humidity'], aggfunc='mean')
````

````python
# --- json_normalize: flatten a nested dictionary ---
data1 = {
    'id': 123,
    'name': 'krish',
    'address': {
        'street': '123 main st',
        'city': 'bangalore',
        'state': 'karnataka'
    }
}
df = pd.json_normalize(data1)
print(df)
print()
print('columns →', list(df.columns))   # address.street, address.city, address.state
````

# Chapter 13 · File Formats & Working With Large Files

| Format | Structure | Storage | Notes |
|---|---|---|---|
| **CSV** | Structured | Row | Human readable, no types, no compression |
| **Excel** (xlsx/xlsb/xlsm) | Structured | Row | Sheets, formatting |
| **JSON** | Semi-structured | — | Nested; how systems exchange data |
| **Parquet** | Structured | **Columnar** | Compressed, binary, read-heavy → analytics |
| **Avro** | Structured | **Row** | Compressed (snappy), write-heavy → streaming |
| **ORC** | Structured | **Columnar** | Compressed, read-heavy → analytics |
| **Word / PDF** | Unstructured | — | Needs parsing before use |

### Why Parquet dominates big data
- **Columnar storage** → faster to locate and scan just the columns you need
- **Compressed** → far less disk space
- **Parallelism** → supports parallel processing
- **Schema evolution** → handles added/removed columns automatically
- **Metadata included** — column names, types, schema (data about data)
- **Binary** → humans can't read it directly (that's the trade-off)

### How data reaches you
- **Email / SFTP ingestion** → download and move the file to a central folder
- **JDBC connection** → url · username · password · port · database · table
- **API** → url · authentication · query string (filters), via the `requests` library


````python
import pandas as pd

df = pd.DataFrame({
    'name': ['alice', 'bob', 'charlie'],
    'age': [25, 30, 35],
    'city': ['newyork', 'sanfrancisco', 'losangeles']
})

df.to_parquet('data.parquet')       # write
newdf = pd.read_parquet('data.parquet')   # read back
newdf
````

````python
# Compare the sizes — parquet compression on a bigger frame
import numpy as np, os
big = pd.DataFrame({'id': range(50000),
                    'city': np.random.choice(['bangalore','chennai','delhi'], 50000),
                    'amount': np.random.randint(100, 9999, 50000)})
big.to_csv('big.csv', index=False)
big.to_parquet('big.parquet')

print('csv     :', round(os.path.getsize('big.csv')/1024, 1), 'KB')
print('parquet :', round(os.path.getsize('big.parquet')/1024, 1), 'KB')
````

### Large files — process in CHUNKS
A 10 GB CSV won't fit in memory. `chunksize` streams it: 1,000,000 rows ÷ 100,000 = 10 chunks processed one at a time.


````python
# Chunked processing: transform and append, writing the header only once
chunk = 10000
first_chunk = True

for part in pd.read_csv('big.csv', chunksize=chunk):
    part['uppercity'] = part['city'].str.upper()        # the transformation
    part.to_csv('newdata.csv', mode='a', index=False, header=first_chunk)
    first_chunk = False                                  # header only on the first write

print('rows written:', len(pd.read_csv('newdata.csv')))
````

````python
# --- Merge many files into one ---
import glob, os
os.makedirs('csvfolder', exist_ok=True)
for i in range(3):                                        # create 3 sample files
    pd.DataFrame({'id':[i*2+1, i*2+2], 'city':['bangalore','chennai']}).to_csv(
        f'csvfolder/file{i}.csv', index=False)

# glob lists all files matching a pattern in a folder
files = glob.glob('csvfolder/*.csv')
print(files)


# List comprehension reads each file; concat stacks them.
# ignore_index=True resets the index on the new dataframe.
mergedf = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
mergedf = mergedf.drop_duplicates()
mergedf.dropna()
mergedf.to_csv('mergedf.csv')
````

````python
# If the files have DIFFERENT columns, use sort=True so concat aligns them
dfs = []
for file in files:
    dfs.append(pd.read_csv(file))
mergedf = pd.concat(dfs, ignore_index=True, sort=True)
mergedf.head()
````

# Chapter 14 · End-to-End Project — Finance ETL

### The brief (a real requirement, written the way a business writes it)
> *A CSV lands in a folder every morning at 12 AM CST from the finance team. Read it into a DataFrame and transform it: remove duplicates; fill nulls with the mean if numeric, `01011900` if a date, `'unknown'` if a string; build the aggregated finance columns. Once cleaned, save the output in parquet format to another folder.*

### Medallion Architecture — how modern pipelines are layered
| Layer | Contents |
|---|---|
| 🥉 **Bronze** | **Raw** data exactly as received — never edited |
| 🥈 **Silver** | **Transformed** — duplicates dropped, nulls fixed, data types corrected |
| 🥇 **Gold** | **Aggregated** business columns — reporting reads from here |

### The finance formulas
| Column | Formula |
|---|---|
| Total revenue | `Revenue` |
| Total expense | `Expenses` |
| Profit | `Revenue − Expenses` |
| Gross margin | `(Revenue − COGS) / Revenue` |
| Operating income | `Revenue − Operating_Expenses` |
| EBITDA | `operating_income + Depreciation + Amortization` |
| Accounts receivable / payable | `AR` / `AP` |
| Year-over-year growth | `Revenue × 0.05` |


````python
# 🥉 BRONZE — generate/receive the raw file
# (This block simulates the finance team's messy CSV. In production you'd
#  simply do: df = pd.read_csv('/path/finance_01012026.csv')
import pandas as pd
import numpy as np
np.random.seed(7)

n = 100
raw = {
    'Date': pd.date_range('2025-03-07', periods=n, freq='h'),
    'Revenue':            np.random.randint(1000, 5000, size=n).astype(float),
    'Expenses':           np.random.randint(500, 2000, size=n).astype(float),
    'COGS':               np.random.randint(300, 1500, size=n).astype(float),
    'Operating_Expenses': np.random.randint(200, 1000, size=n).astype(float),
    'Depreciation':       np.random.randint(50, 200, size=n),
    'Amortization':       np.random.randint(50, 200, size=n),
    'Cash_Inflows':       np.random.randint(1000, 3000, size=n),
    'Cash_Outflows':      np.random.randint(500, 2500, size=n),
    'AR':                 np.random.randint(100, 1500, size=n),
    'AP':                 np.random.randint(100, 1500, size=n),
}
dfi = pd.DataFrame(raw)

# inject 5 REAL duplicate rows (an exact copy of the first five)
dfi = pd.concat([dfi, dfi.head(5)], ignore_index=True)

# inject nulls into the numeric columns
null_rows = np.random.choice(dfi.index, size=15, replace=False)
for c in ['Revenue', 'Expenses', 'Operating_Expenses', 'COGS']:
    dfi.loc[null_rows, c] = np.nan

# a string column, with some nulls
dfi['Department'] = (['Sales', 'Marketing', 'Finance', 'HR', 'Legal'] * 21)[:len(dfi)]
dfi.loc[10:20, 'Department'] = np.nan

dfi.to_csv('finance_raw.csv', index=False)
print('🥉 Bronze written: finance_raw.csv —', len(dfi), 'rows')
````

````python
# --- EXTRACT: the actual starting point of the project ---
df = pd.read_csv('finance_raw.csv')
df.info()
````

````python
# 🥈 SILVER — transformation step 1: drop duplicates
print('rows before:', len(df))
df.drop_duplicates(inplace=True)
print('rows after :', len(df))
````

````python
df['Date'] = df['Date'].fillna('01011900')
df['Department'] = df['Department'].fillna('unknown')
````

````python
# 🥈 SILVER — step 2: fill nulls by column TYPE
# numeric → mean
for col in ['Revenue', 'Expenses', 'COGS', 'Operating_Expenses']:
    df[col] = df[col].fillna(df[col].mean())

# date → business sentinel
df['Date'] = df['Date'].fillna('01011900')

# string → 'unknown'
df['Department'] = df['Department'].fillna('unknown')

df.isnull().sum()
df.info()
````

````python
# 🥇 GOLD — build the aggregated reporting columns
df['total_revenue']         = df['Revenue']
df['total_expense']         = df['Expenses']
df['profit']                = df['Revenue'] - df['Expenses']
df['gross_margin']          = (df['Revenue'] - df['COGS']) / df['Revenue']
df['operating_income']      = df['Revenue'] - df['Operating_Expenses']
df['EBITDA']                = df['operating_income'] + df['Depreciation'] + df['Amortization']
df['accounts_receivables']  = df['AR']
df['accounts_payables']     = df['AP']
df['year_over_year_growth'] = df['Revenue'] * 0.05

df[['total_revenue','profit','gross_margin','operating_income','EBITDA']].head()
````

````python
# --- LOAD: save the gold layer in parquet (and csv for inspection) ---
df.to_parquet('finance_gold.parquet')
df.to_csv('finance_gold.csv', index=False)

# post-load validation — always read it back and confirm
check = pd.read_parquet('finance_gold.parquet')
print('rows :', len(check))
print('cols :', len(check.columns))
print('nulls:', check.isnull().sum().sum())
check.head(3)
````

````python
# A quick business summary — what the dashboard would show
df.groupby('Department').agg(
    total_revenue=('total_revenue', 'sum'),
    total_profit=('profit', 'sum'),
    avg_gross_margin=('gross_margin', 'mean')
).round(2)
````
